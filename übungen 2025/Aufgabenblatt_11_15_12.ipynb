1. Die Gewerbeaufsicht will die Schankgenauigkeit auf dem Oktoberfest überprüfen.
Bei einer Stichprobe n=9 ergibt sich als durchschnittlicher Inhalt pro Maß x = 0.97 Liter.
Es wird angenommen, dass der Inhalt pro Maß normalverteilt ist.
(a) Berechnen Sie ein Konfidenzintervall zu 1−α = 0.95 für den unbekannten Erwartungswert
µ unter der Annahme σ
2 = 0.0752.
(b) Berechnen Sie ein Konfidenzintervall zu 1−α = 0.95 für den unbekannten Erwartungswert
µ, wenn aus den Daten s
2 = 0.0752 bestimmt wurde.
(c) Welches Intervall ist kleiner? Begründen Sie Ihre Antwort.


```python
from scipy.stats import norm
import math
# a
x_bar = 0.97
n = 9
varianz = 0.0752
abweichung = math.sqrt(varianz)/math.sqrt(n)
eins_minus_a = 0.95
alpha = 1 - eins_minus_a
print("Konfidenzintervall: [", x_bar - norm.ppf(1-alpha/2)*abweichung, ", ", x_bar + norm.ppf(1-alpha/2)*abweichung, "]")
```

```python
# b
from scipy.stats import t
x_bar = 0.97
n = 9
s2 = 0.0752
standardfehler = math.sqrt(varianz)/math.sqrt(n)
eins_minus_a = 0.95
alpha = 1 - eins_minus_a
df = n - 1
print("Konfidenzintervall: [", x_bar - t.ppf(1-alpha/2, df)*abweichung, ", ", x_bar + t.ppf(1-alpha/2, df)*abweichung, "]")
```

2. Es seien n Zufallsvariablen Xi (i = 1,2,...,n) unabhängig und identisch verteilt mit dem
Mittelwert µ und der Varianz σ
2 = 36.
Wie groß muss n sein, damit die Wahrscheinlichkeit dafür, dass X von µ um nicht mehr
als 0,5 abweicht, gleich 0,99 ist?


```python
import math
from scipy.stats import norm

sigma = 6.0
epsilon = 0.5
confidence = 0.99
alpha = 1 - confidence

z = norm.ppf(1 - alpha/2)
n_required = (z * sigma / epsilon) ** 2

z, n_required
```

3. Das in der Schmuckherstellung verarbeitete Gold besteht, je nach Wert der Legierung,
aus weiteren, hinzugefügten Metallen wie beispielsweise Nickel oder Kupfer. Bei einer
Stichprobe von 116 Kontrollmessungen, kam es zu folgenden Werten für den Gehalt an
Nickel, gemessen in ml/l. Der Stichprobenmittelwert ergab: 25,452 und die Stichprobenvarianz:
0,7225.
Die Aufgabe besteht nun darin, das 96%-igen Konfidenzintervall für den mittleren Nickelgehalt
innerhalb der Herstellung zu ermitteln.


```python
import math 
stichprobenmittelwert = 25.452
n=116
sigma=math.sqrt(0.7225)/math.sqrt(n)
konfidenz = 0.96
alpha=1-konfidenz
intervall_unten = stichprobenmittelwert - norm.ppf(1-alpha/2) * sigma
intervall_oben = stichprobenmittelwert +norm.ppf(1-alpha/2) * sigma
print("Konfidenzintervall: [", intervall_unten, ", " , intervall_oben,"]")
```
