2. Eine Telefonauskunft wird durchschnittlich 12 mal pro Stunde angewählt.
Mit welcher Wahrscheinlichkeit erhält die Auskunft
(a) in den nächsten 5 Minuten keinen Anruf,
(b) innerhalb von 10 Minuten mindestens 5 Anrufe,
(c) innerhalb von 20 Minuten höchstens 6 Anrufe?

```python
from scipy.stats import poisson
print("Aufgabe (a):", poisson.pmf(0, 12/60*5))
print("Aufgabe (b):", 1 - poisson.cdf(4, 12/60*10))
print("Aufgabe (c):", poisson.cdf(6, 12/60*20))
```

3. Eine Serienproduktion von Transistoren hat einen gleichbleibenden Ausschuss-Anteil von
p=3%.
Wie groß ist die Wahrscheinlichkeit, dass in einer Stichprobe von n=50 entnommenen
Einheiten
(a) keine
(b) genau zwei
(c) höchstens zwei
fehlerhafte Bauteile sind?
Wie groß sind Erwartungswert und Varianz der Anzahl X der fehlerhaften Bauteile in der
Stichprobe?

```python
from scipy.stats import binom
print("Aufgabe (a):", binom.pmf(0, 50, 0.03))
print("Aufgabe (b):", binom.pmf(2, 50, 0.03))
print("Aufgabe (c):", binom.cdf(2, 50, 0.03))
print("Erwartungswert:", 50 * 0.03)
print("Varianz:", 50 * 0.03 * (1 - 0.03))
```
