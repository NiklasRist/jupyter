Eine stochastisch unabhängige, Stichprobe der Länge n=8 ergibt folgende Daten:
i 1 2 3 4 5 6 7 8
xi 20,7 17,3 26,4 40,9 20,4 25,7 27,3 18,0
(a) Bestimmen Sie das Konfidenzintervall des Mittelwertes zur Irrtumswahrscheinlichkeit
α = 0, 1
(b) Bestimmen Sie das Konfidenzintervall der Varianz zur Irrtumswahrscheinlichkeit
α = 0, 05


```python
import numpy as np
from scipy.stats import t, chi2
import math
n=8
x_i =  np.array([20.7, 17.3, 26.4, 40.9, 20.4, 25.7, 27.3, 18.0])
mü = x_i.sum()/n
varianz = 1/(n-1) * ((x_i-mü)**2).sum()

# a
alpha = 0.1
unteres_intervall = mü - t.ppf(1-alpha/2, n-1)*math.sqrt(varianz)/math.sqrt(n)
oberes_intervall = mü + t.ppf(1-alpha/2, n-1)*math.sqrt(varianz)/math.sqrt(n)
print("Konfindenzintervall des Mittelwerts: [", unteres_intervall, oberes_intervall, "]")

# b 
alpha = 0.05
unteres_intervall = (n-1)*varianz/chi2.ppf(1-alpha/2 , n-1)
oberes_intervall = (n-1)*varianz/chi2.ppf(alpha/2 , n-1)
print("Konfindenzintervall der Varianz: [", unteres_intervall, oberes_intervall, "]")
```
