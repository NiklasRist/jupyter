2. Eine Glühbirne in einer Notbeleuchtung ist ununterbrochen in Betrieb bis sie ausfällt. Die
Zufallsvariable X, durch die die Lebensdauer (in Stunden) von Glühbirnen modelliert
wird, sei exponentialverteilt mit Parameter 
λ = 1 / 6000 = 6000^−1
.
(a) Wie groß ist die Wahrscheinlichkeit, dass eine Glühbirne mindestens 6000 Stunden
funktioniert?
(b) Eine Glühbirne sei bereits 3000 Stunden in Betrieb. Wie groß ist die bedingte Wahrscheinlichkeit,dass sie noch weitere 3000 Stunden nicht ausfällt?


```python
from scipy.stats import expon
print("Aufgabe (a):", 1-expon.cdf(1/6000))
num = 1 - expon.cdf(6000, scale=6000)
den = 1 - expon.cdf(3000, scale=6000)
print("Aufgabe (b):", num/den)
```

3. Ein Unternehmen produziert Vitamin-C-Tabletten, die jeweils 100 mg Vitamin-C enthalten sollen. Produktionsbedingt ist mit einer Standardabweichung von 5 mg zu rechnen. Ausgeliefert werden Tabletten mit einem Vitamin-C-Gehalt der zwischen 95 mg und 110 mg liegt. Berechnen Sie die Wahrscheinlichkeit, dass bei einer zufälligen Entnahme einer Tablette diese im für die Auslieferung erlaubten Bereich liegt.

```python
from scipy.stats import norm
mu = 100
sigma = 5
print("Wahrscheinlichkeit: ", norm.cdf(110, loc=mu, scale=sigma) - norm.cdf(95, loc=mu, scale=sigma))
```
