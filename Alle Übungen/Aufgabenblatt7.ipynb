## Aufgabe 1  

![img](./Bilder/Glücksrad2.png)

Ein Spieler zahlt 1 €, um einmal an dem Glücksrad drehen zu dürfen. Er gewinnt den
doppelten Betrag der dunklen Zahlen und verliert den einfachen Betrag der hellen Zahlen
in €.

(a) Stellen Sie eine Wertetabelle für den Gewinn X auf.

```python
# Aufgabe 1: Lösung
# Dunkle Zahlen gemäß Abbildung (bitte Kontrolle der Liste)
dark = [2, 5, 7, 10, 15, 18]
values = list(range(1, 21))
def X(i, stake=1):
    return (2*i - stake) if i in dark else (-i - stake)

# Wertetabelle
table = [(i, X(i)) for i in values]
df1 = pd.DataFrame(table, columns=['i', 'X(i)'])
display(df1)





```

(b) Berechnen Sie den Erwartungswert E(X).

```python
# Erwartungswert für Stake=1
p = 1/20
E1 = np.mean([X(i) for i in values])
print(f"E(X) bei Einsatz 1€: {E1:.3f} €")
```

(c) Korrigieren Sie den Einsatz so, dass das Spiel fair wird.

```python
# Fairer Einsatz s0 so dass E(X)=0
s0 = sum((2*i if i in dark else -i) for i in values)/20
print(f"Fairer Einsatz: {s0:.3f} €")
```

(d) Berechnen Sie für beide Fälle die Varianz und die Standardabweichung.

```python
def var_and_std(stake):
    xs = [X(i, stake) for i in values]
    return np.var(xs, ddof=0), np.std(xs, ddof=0)

var1, std1 = var_and_std(1)
var0, std0 = var_and_std(s0)
print(f"Varianz bei Einsatz 1€: {var1:.3f}, SD: {std1:.3f}")
print(f"Varianz bei fairem Einsatz {s0:.3f}€: {var0:.3f}, SD: {std0:.3f}")
```

## Aufgabe 2  
Bei einer Produktion können pro Tag folgende Anzahl an Störfällen mit den entsprechenden Wahrscheinlichkeiten auftreten:

| Anzahl | 0   | 1    | 2    | 3    |
|--------|-----|------|------|------|
| p      | 0,6 | 0,12 | 0,2  | 0,08 |

Die Kosten (in tsd Euro) für die Behebung der einzelnen Störfälle kann durch folgende Funktion ermittelt werden:

$$K(x) = 6 − \frac{6}{(1 + x)}$$

Was sind die zu erwartenden Kosten pro Tag?

Verwenden Sie folgenden Satz: Ist X eine beliebige Zufallsvariable und g(x) eine beliebige reelle Funktion. Dann ist:

$$
E\bigl(g(X)\bigr)
=
\begin{cases}
\displaystyle \sum_i g(x_i)\,p_i, & \text{für \(X\) diskret},\\[6pt]
\displaystyle \int_{-\infty}^{\infty} g(x)\,f(x)\,\mathrm{d}x, & \text{für \(X\) stetig}.
\end{cases}
$$






```python
# Aufgabe 2: Lösung
x = np.array([0, 1, 2, 3])
p = np.array([0.6, 0.12, 0.2, 0.08])
K = lambda x: 6 - 6 / (1 + x)
E_cost = np.sum(K(x) * p)
print(f"Erwartete Kosten pro Tag: {E_cost:.3f} tsd Euro")
```

## Aufgabe 3  
Eine Serienproduktion von Transistoren hat einen gleichbleibenden Ausschuss-Anteil von p = 3 %.
Wie groß ist die Wahrscheinlichkeit, dass in einer Stichprobe von n = 50 entnommenen Einheiten

(a) keine  
fehlerhafte Bauteile sind?  



```python
n, p = 50, 0.03
rv = binom(n, p)
prob_none = rv.pmf(0)
print(f"P(X=0) = {prob_none:.5f}")
```

(b) genau zwei
fehlerhafte Bauteile sind?  

```python
prob_two = rv.pmf(2)
prob_at_most_two = rv.cdf(2)
```

(c) höchstens zwei  
fehlerhafte Bauteile sind?  
Wie groß sind Erwartungswert und Varianz der Anzahl X der fehlerhaften Bauteile in der Stichprobe?

```python
prob_at_most_two = rv.cdf(2)
print(f"P(X ≤ 2) = {prob_at_most_two:.5f}")

mean = rv.mean()
var = rv.var()
print(f"Erwartungswert: {mean:.3f}")
print(f"Varianz: {var:.3f}")
```
