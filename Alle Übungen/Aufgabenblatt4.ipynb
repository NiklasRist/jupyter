## Aufgabe 1

Eine Firma lässt zwei verschiedene Werbespots (W1, W2) im Fernsehen senden. Es ist bekannt, dass ein Fernsehzuschauer den Werbespot W1 mit einer Wahrscheinlichkeit von 0,1 und den Werbespot W2 mit Wahrscheinlichkeit 0,15 und beide Werbespots mit Wahrscheinlichkeit 0,05 sieht.  
Berechnen Sie die Wahrscheinlichkeit dafür, dass eine Person  
- genau einen  
- mindestens einen  
- keinen  
- höchstens einen 
 
Werbespot sieht.


```python
p_w1 = 0.1
p_w2 = 0.15
p_both = 0.05

p_exactly_one = (p_w1 - p_both) + (p_w2 - p_both)
p_at_least_one = p_w1 + p_w2 - p_both
p_none = 1 - p_at_least_one
p_at_most_one = 1 - p_both

print(f"P(exakt ein Werbespot) = {p_exactly_one:.4f}")
print(f"P(mindestens ein Werbespot) = {p_at_least_one:.4f}")
print(f"P(kein Werbespot) = {p_none:.4f}")
print(f"P(höchstens ein Werbespot) = {p_at_most_one:.4f}")
```

## Aufgabe 2

Gegeben ist die quadratische Gleichung  
$$x^2 + p\,x + q = 0$$  
wobei p und q Zufallszahlen im Intervall [0;4] sind.  
Gesucht ist die Wahrscheinlichkeit, dass die quadratische Gleichung mindestens eine reelle Lösung hat.  
Hinweis: Skizzieren Sie im Bereich mit $0 \le q \le 4$; $0 \le p \le 4$ den Graph der Funktion  
$$q = \frac{p^2}{4}$$  
und markieren Sie den Bereich mit  
$$q \le \frac{p^2}{4}$$

```python
# Numerische Berechnung der Fläche und Wahrscheinlichkeit
p = np.linspace(0, 4, 10001)
# Fläche unter q = p^2/4
area = np.trapz(p**2 / 4, p)
prob = area / 16  # Gesamtfläche 4*4 = 16
print(f"Wahrscheinlichkeit für mindestens eine reelle Lösung: {prob:.4f}")

# Visualisierung des Bereichs
P, Q = np.meshgrid(p, p)
region = Q <= (P**2 / 4)
plt.figure()
plt.contourf(P, Q, region, levels=[-0.5, 0.5, 1.5], alpha=0.5)
plt.plot(p, p**2 / 4, 'k')
plt.xlabel('p')
plt.ylabel('q')
plt.title('Bereich mit mindestens einer reellen Lösung')
plt.show()
```

## Aufgabe 3

Ein Glücksrad mit den Mittelpunktwinkeln $60^\circ$, $120^\circ$ und $180^\circ$ ist mit den Zahlen 20, 10 und 6 beschriftet.  
![img](./Bilder/Glücksrad.png)

Es wird zweimal gedreht.

(a) Wie groß ist die Wahrscheinlichkeit, dass die Summe der erhaltenen Zahlen genau 30 ergibt?  


```python
outcomes = [20, 10, 6]
probs = [60/360, 120/360, 180/360]  

pairs = [(x, y) for x in outcomes for y in outcomes]
pair_probs = {pair: probs[outcomes.index(pair[0])] * probs[outcomes.index(pair[1])] for pair in pairs}

p_sum_30 = sum(p for (xy, p) in zip(pair_probs.keys(), pair_probs.values()) if sum(xy) == 30)
print(f"P(Summe = 30) = {p_sum_30:.4f}")
```

(b) Wie groß ist die Wahrscheinlichkeit, dass die Summe größer als 12 ist?  

```python
p_sum_gt12 = sum(p for (xy, p) in pair_probs.items() if sum(xy) > 12)
print(f"P(Summe > 12) = {p_sum_gt12:.4f}")
```

(c) Mit welcher Wahrscheinlichkeit ist die Summe kleiner als 30?

```python
p_sum_lt30 = sum(p for (xy, p) in pair_probs.items() if sum(xy) < 30)
print(f"P(Summe < 30) = {p_sum_lt30:.4f}")
```
