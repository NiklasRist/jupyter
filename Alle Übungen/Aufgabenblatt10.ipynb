```python
import scipy.stats as st
import numpy as np
import sympy as sp
```

## Aufgabe 1

Eine Maschine verpackt Schrauben. In einer Packung sind 80 Schrauben. Das Gewicht der Schrauben ist annährend normalverteilt mit dem Mittelwert µ = 4g und der Standardabweichung σ=0,1g. Bestimmen Sie für das Nettogewicht X der Packung die Wahrscheinlichkeit P(X≤ 310g)

```python
# Parameter pro Schraube
mu = 4  # g
sigma = 0.1  # g
n = 80  # Anzahl Schrauben

# Verteilung der Summe X ~ N(n*mu, sqrt(n)*sigma)
mu_X = n * mu
sigma_X = sigma * (n ** 0.5)

prob = st.norm.cdf(310, loc=mu_X, scale=sigma_X)
print(f"P(X ≤ 310g) = {prob:.6f}")
```

## Aufgabe 2
Im Wasserwerk soll der Salzgehalt des Trinkwassers (in mg/l) bestimmt werden. Die Messung ist fehleranfällig, daher wird mehrmals gemessen und dann das Stichprobenmittel berechnet. Es kann davon ausgegangen werden, dass der tatsächliche Salzgehalt bei allen Messungen derselbe ist. Ein systematischer Fehler ist auszuschließen. Erfahrungsgemäß ist die Standardabweichung der Messungen 1mg/l.

(a) Wie groß ist die Wahrscheinlichkeit, dass das Stichprobenmittel um mehr als 0.1mg/l vom zu schätzenden Erwartungswert abweicht, falls 100 Messungen durchgeführt wurden?

```python
sigma = 1  # mg/l
n = 100

# Standardabweichung des Stichprobenmittels
sigma_bar = sigma / (n ** 0.5)

# Wahrscheinlichkeit P(|X̄ - μ| > 0.1)
prob_a = 2 * (1 - st.norm.cdf(0.1, loc=0, scale=sigma_bar))
print(f"P(|X̄ - μ| > 0.1) = {prob_a:.6f}")
```

(b) Wieviele Messungen müssen mindestens durchgeführt werden, so dass die Wahrscheinlichkeit, dass das Stichprobenmittel der Messungen um mehr als 0,1mg/l vom wahren Wert abweicht, höchstens 5% beträgt?

Hinweis: Verwenden Sie die Wahrscheinlichkeiten der σ-Intervalle der Normalverteilung.


```python
sigma = 1  # mg/l
alpha = 0.05

# z-Wert für zweiseitiges Konfidenzintervall
z = st.norm.ppf(1 - alpha / 2)

# n = (z * sigma / 0.1)^2
n_required = int(np.ceil((z * sigma / 0.1) ** 2))
print(f"Minimale Stichprobengröße n = {n_required}")
```

## Aufgabe 3 
Überprüfen Sie die folgenden Schätzfunktionen, ob sie erwartungstreu für µ sind

- $$T_1 = \sum_{i=1}^n X_i - n \bar{X} + X_1$$
- $$T_2 = 0.2 X_1 + 0.9 X_2 - 0.1 X_3$$
- $$T_3 = \frac{1}{n - 2} \left(\sum_{i=1}^{n-1} X_i + 2 X_1\right)$$

```python
# Symbole
n, mu = sp.symbols('n mu')

# Erwartungswerte berechnen
E_T1 = mu  # E[T1] = E[X1] = mu
E_T2 = (0.2 + 0.9 - 0.1) * mu  # E[T2] = mu
E_T3 = ((n - 1) * mu + 2 * mu) / (n - 2)  # E[T3]

print(f"E[T1] = {E_T1}")
print(f"E[T2] = {E_T2}")
print(f"E[T3] = {sp.simplify(E_T3)}")
```
