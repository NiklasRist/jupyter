## Aufgabe 1

Eine Firma produziert insgesamt 350 elektronische Bauteile des gleichen Typs. Aus Erfahrung weiß man dass davon jedes 70 te defekt ist.
Um die Qualität zu prüfen untersucht der Käufer der Bauteile eine Stichprobe. Dazu werden zufällig 20 Bauteile heraus genommen und untersucht. Wenn mehr als ein defektes Bauteil gefunden wird, wird die Sendung zurückgeschickt.
Wie groß ist die Wahrscheinlichkeit, dass die Warensendung wieder zurück geschickt wird?

```python

N = 350
K = N // 70  # Anzahl defekter Bauteile
n = 20
p_return = sum(math.comb(K, k) * math.comb(N-K, n-k) for k in range(2, K+1)) / math.comb(N, n)
print(f"Wahrscheinlichkeit, dass die Sendung zurückgeschickt wird: {p_return:.4f}")
```

## Aufgabe 2

Eine Telefonauskunft wird durchschnittlich 12 mal pro Stunde angewählt.
Mit welcher Wahrscheinlichkeit erhält die Auskunft

(a) in den nächsten 5 Minuten keinen Anruf,

```python
lambda_hour = 12
lam5 = lambda_hour * 5/60
p0 = math.exp(-lam5)
print(f"(a) P(kein Anruf in 5 Minuten) = {p0:.4f}")
```

(b) innerhalb von 10 Minuten mindestens 5 Anrufe,

```python
lam10 = lambda_hour * 10/60
p_ge5 = 1 - sum(math.exp(-lam10) * lam10**k / math.factorial(k) for k in range(0, 5))
print(f"(b) P(mindestens 5 Anrufe in 10 Minuten) = {p_ge5:.4f}")
```

(c) innerhalb von 20 Minuten höchstens 6 Anrufe?

```python
lam20 = lambda_hour * 20/60
p_le6 = sum(math.exp(-lam20) * lam20**k / math.factorial(k) for k in range(0, 7))
print(f"(c) P(höchstens 6 Anrufe in 20 Minuten) = {p_le6:.4f}")
```

## Aufgabe 3

Eine Serienproduktion von Transistoren hat einen gleichbleibenden Ausschuss-Anteil von p=3%.
Wie groß ist die Wahrscheinlichkeit, dass in einer Stichprobe von n=50 entnommenen Einheiten

(a) keine

```python
n3 = 50
p = 0.03
p0 = math.comb(n3, 0) * p**0 * (1-p)**(n3-0)
print(f"(a) P(keine fehlerhaften Bauteile) = {p0:.4f}")
```

(b) genau zwei
fehlerhafte Bauteile sind?

```python
p2 = math.comb(n3, 2) * p**2 * (1-p)**(n3-2)
print(f"(b) P(genau zwei fehlerhafte Bauteile) = {p2:.4f}")
```

(c) höchstens zwei
fehlerhafte Bauteile sind?
Wie groß sind Erwartungswert und Varianz der Anzahl X der fehlerhaften Bauteile in der Stichprobe?

```python
p_le2 = sum(math.comb(n3, k) * p**k * (1-p)**(n3-k) for k in range(0, 3))
E = n3 * p
Var = n3 * p * (1-p)
print(f"(c) P(höchstens zwei fehlerhafte Bauteile) = {p_le2:.4f}")
print(f"Erwartungswert E[X] = {E:.2f}")
print(f"Varianz Var[X] = {Var:.2f}")
```
