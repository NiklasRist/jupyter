## Aufgabe 1
In einer Gruppe von 900 Personen haben sich 600 prophylaktisch gegen Grippe impfen
lassen. Nach einer bestimmten Zeit wurde jedes Gruppenmitglied danach befragt, wer an
einer Grippe erkrankte. Die Ergebnisse werden in einer 4-Feldtafel dargestellt. 

| Gruppe        | B erkrankt | $\overline{B}$ gesund | Summe |
|---------------|------------|----------|-------|
| A geimpft     | 60         | 540      | 600   |
| $\overline{A}$ ungeimpft   | 120        | 180      | 300   |
| Summe         | 180        | 720      | 900   |

Berechnen Sie: P(A); P(B); P(A∩B); P(B|A); P(A|B); P($\overline{A}$∩B); P(B|$\overline{A}$)

```python
n_total = 900
n_A = 600
n_B = 180
n_A_and_B = 60

P_A = n_A / n_total
P_B = n_B / n_total
P_A_and_B = n_A_and_B / n_total
P_B_given_A = n_A_and_B / n_A
P_A_given_B = n_A_and_B / n_B

print(f"P(A) = {P_A:.4f}")
print(f"P(B) = {P_B:.4f}")
print(f"P(A ∩ B) = {P_A_and_B:.4f}")
print(f"P(B|A) = {P_B_given_A:.4f}")
print(f"P(A|B) = {P_A_given_B:.4f}")
```

## Aufgabe 2
In einem Großversuch wurde ein Medikament getestet. Die Ergebnisse sind in einer Tabelle festgehalten. Dabei bedeuten:

M... Medikament genommen  
$\overline{M}$... Placebo genommen  
G... Gesund geworden  
$\overline{G}$... nicht gesund geworden  

|               | G    | $\overline{G}$ | Summe |
|---------------|------|---------------|-------|
| M             | 6312 | 87            | 6399  |
| $\overline{M}$ | 312  | 4390          | 4702  |
| Summe         | 6624 | 4477          | 11101 |

(a) Stellen Sie die dazugehörigen Baumdiagramme auf.

```python
# Aufgabe 2(a): Zweigwahrscheinlichkeiten berechnen
total = 11101
n_M = 6399
n_notM = 4702
n_M_and_G = 6312
n_M_and_notG = 87
n_notM_and_G = 312
n_notM_and_notG = 4390

P_M = n_M / total
P_notM = n_notM / total
P_G_given_M = n_M_and_G / n_M
P_notG_given_M = n_M_and_notG / n_M
P_G_given_notM = n_notM_and_G / n_notM
P_notG_given_notM = n_notM_and_notG / n_notM

print(f"P(Medikament) = {P_M:.4f}")
print(f"P(Placebo) = {P_notM:.4f}")
print(f"P(G | M) = {P_G_given_M:.4f}")
print(f"P(¬G | M) = {P_notG_given_M:.4f}")
print(f"P(G | ¬M) = {P_G_given_notM:.4f}")
print(f"P(¬G | ¬M) = {P_notG_given_notM:.4f}")
```

(b) Wie groß ist die Wahrscheinlichkeit bei einer Person, von der man weiß, dass sie das Medikament eingenommen hat, zu genesen?

```python
print(f"P(G | Medikament) = {P_G_given_M:.4f}")
```

(c) Wie groß ist die Wahrscheinlichkeit bei einer Person, von der man weiß, dass sie das Placebo eingenommen hat, nicht zu genesen?

```python
print(f"P(¬G | Placebo) = {P_notG_given_notM:.4f}")
```

## Aufgabe 3
Bei einer Sportveranstaltung wird ein Dopingtest durchgeführt. Wenn ein Sportler gedopt
hat, dann fällt der Test zu 99% positiv aus. Hat ein Sportler aber nicht gedopt, zeigt der
Test trotzdem zu 5% ein positives Ergebnis an. Aus Erfahrung weiß man, dass 20% der
Sportler gedopt sind.

(a) Bestimmen Sie die Wahrscheinlichkeit, dass eine Dopingprobe positiv ausfällt.

```python
P_D = 0.20  
P_notD = 1 - P_D
P_pos_given_D = 0.99
P_pos_given_notD = 0.05
P_pos = P_pos_given_D * P_D + P_pos_given_notD * P_notD
print(f"P(Positiv) = {P_pos:.4f}")
```

(b) Bestimmen Sie die Wahrscheinlichkeit, dass der Test negativ ausfällt, obwohl der Sportler gedopt hat.

```python
P_not_pos_given_D = 1 - P_pos_given_D
print(f"P(Negativ trotz Doping) = {P_not_pos_given_D:.4f}")
```

(c) Berechnen Sie die Wahrscheinlichkeit, dass ein Sportler gedopt hat, wenn seine Dopingprobe negativ war

```python
P_not_pos = 1 - P_pos
P_D_given_not_pos = (P_not_pos_given_D * P_D) / P_not_pos
print(f"P(Doping | Test negativ) = {P_D_given_not_pos:.4f}")
```
