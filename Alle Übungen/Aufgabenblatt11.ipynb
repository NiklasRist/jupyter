## Aufgabe 1

Die Gewerbeaufsicht will die Schankgenauigkeit auf dem Oktoberfest überprüfen.
Bei einer Stichprobe n=9 ergibt sich als durchschnittlicher Inhalt pro Maß $\bar{x} = 0.97$ Liter.
Es wird angenommen, dass der Inhalt pro Maß normalverteilt ist.

(a) Berechnen Sie ein Konfidenzintervall zu $1-\alpha = 0.95$ für den unbekannten Erwartungswert $\mu$ unter der Annahme $\sigma^2 = 0.0752$.

```python
alpha = 0.05
n = 9
x_bar = 0.97
sigma = math.sqrt(0.0752)
z = st.norm.ppf(1 - alpha/2)
se = sigma / math.sqrt(n)
ci_lower = x_bar - z * se
ci_upper = x_bar + z * se
ci_lower, ci_upper
```

(b) Berechnen Sie ein Konfidenzintervall zu $1-\alpha = 0.95$ für den unbekannten Erwartungswert $\mu$, wenn aus den Daten $s^2 = 0.0752$ bestimmt wurde.

```python
s2 = 0.0752
s = math.sqrt(s2)
df = n - 1
t = st.t.ppf(1 - alpha/2, df)
se_t = s / math.sqrt(n)
ci_t_lower = x_bar - t * se_t
ci_t_upper = x_bar + t * se_t
ci_t_lower, ci_t_upper
```

(c) Welches Intervall ist kleiner? Begründen Sie Ihre Antwort.

```python
# Vergleich der Intervalllängen aus (a) und (b)
length1 = 2 * z * (sigma / math.sqrt(n))
length2 = 2 * t * (s / math.sqrt(n))
length1, length2
```

## Aufgabe 2

Es seien $n$ Zufallsvariablen $X_i$ (i = 1,2,...,n) unabhängig und identisch verteilt mit dem Mittelwert $\mu$ und der Varianz $\sigma^2 = 36$. Wie groß muss $n$ sein, dass die Wahrscheinlichkeit dafür, dass $\bar X$ um nicht mehr als 0,5 von $\mu$ abweicht, gleich 0,99 ist?

```python
sigma2 = 36
epsilon = 0.5
prob = 0.99
# Chebyshev-Ungleichung: 1 - sigma2/(n*epsilon^2) = prob
n_required = math.ceil(sigma2 / (epsilon**2 * (1 - prob)))
n_required
```

## Aufgabe 3

Das in der Schmuckherstellung verarbeitete Gold besteht, je nach Wert der Legierung, aus weiteren, hinzugefügten Metallen wie beispielsweise Nickel oder Kupfer. Bei einer Stichprobe von 116 Kontrollmessungen kam es zu folgenden Werten für den Gehalt an Nickel, gemessen in ml/l. Der Stichprobenmittelwert ergab: 25,452 und die Stichprobenvarianz: 0,7225.
Die Aufgabe besteht nun darin, das 96%-ige Konfidenzintervall für den mittleren Nickelgehalt innerhalb der Herstellung zu ermitteln.

```python
alpha = 0.04
n = 116
x_bar = 25.452
s2 = 0.7225
s = math.sqrt(s2)
df = n - 1
t = st.t.ppf(1 - alpha/2, df)
se = s / math.sqrt(n)
ci_lower = x_bar - t * se
ci_upper = x_bar + t * se
ci_lower, ci_upper
```
