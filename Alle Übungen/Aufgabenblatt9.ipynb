## Aufgabe 1
Der Bus fährt alle 10 Minuten. Kennt man den genauen Abfahrtsplan nicht, so geht man einfach zu einer gewissen Zeit an die Haltestelle. Die Wartezeit an der Haltestelle folgt dann einer stetigen Gleichverteilung zwischen 0 und 10 Minuten. Die Zufallsvariable ist die Zeit in Minuten, die bis zum nächsten Bus vergeht. Dabei ist jede reelle Zahl im Intervall von [0;10] als Wartezeit möglich und alle Wartezeiten haben die gleiche Wahrscheinlichkeit.

(a) Wie lange muss man durchschnittlich warten?

```python
a, b = 0, 10
E = (a + b) / 2
E
```

(b) Wie ist die Wahrscheinlichkeit länger als 6 Minuten zu warten?

```python
a, b, x = 0, 10, 6
P = (b - x) / (b - a)
P
```

## Aufgabe 2
Eine Glühbirne in einer Notbeleuchtung ist ununterbrochen in Betrieb bis sie ausfällt. Die Zufallsvariable X, durch die die Lebensdauer (in Stunden) von Glühbirnen modelliert wird, sei exponentialverteilt mit Parameter $$λ = \frac{1}{6000} = 6000^{-1}$$.

(a) Wie groß ist die Wahrscheinlichkeit, dass eine Glühbirne mindestens 6000 Stunden funktioniert?

```python
lam = 1/6000
# P(X >= 6000) = exp(-λ * 6000)
P1 = math.exp(-lam * 6000)
P1
```

(b) Eine Glühbirne sei bereits 3000 Stunden in Betrieb. Wie groß ist die bedingte Wahrscheinlichkeit, dass sie noch weitere 3000 Stunden nicht ausfällt?

```python
lam = 1/6000
# Gedächtnislosigkeit der Exponentialverteilung:
# P(X >= 6000 | X >= 3000) = P(X >= 3000)
P2 = math.exp(-lam * 3000)
P2
```

## Aufgabe 3
Ein Unternehmen produziert Vitamin-C-Tabletten, die jeweils 100 mg Vitamin-C enthalten sollen. Produktionsbedingt ist mit einer Standardabweichung von 5 mg zu rechnen. Ausgeliefert werden Tabletten mit einem Vitamin-C-Gehalt der zwischen 95 mg und 110 mg liegt. Berechnen Sie die Wahrscheinlichkeit, dass bei einer zufälligen Entnahme einer Tablette diese im für die Auslieferung erlaubten Bereich liegt.

```python
mu, sigma = 100, 5
def cdf(x):
    return 0.5 * (1 + math.erf((x - mu) / (sigma * math.sqrt(2))))
# P(95 <= X <= 110)
P3 = cdf(110) - cdf(95)
P3
```
