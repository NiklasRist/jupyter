## Aufgabe 1
Gegeben ist die folgende Urliste  
1, 2, 1, 2, 10, 10, 20, 1, 2, 1  
Berechnen Sie die folgenden Quantile:  
Q0.2, Q0.25, Q0.4, Q0.5, Q0.6, Q0.75, Q0.99

```python
import numpy as np
import matplotlib.pyplot as plt

urliste = [1, 2, 1, 2, 10, 10, 20, 1, 2, 1]
quantile = np.percentile(urliste, [20, 25, 40, 50, 60, 75, 99])
print(quantile)
```

## Aufgabe 3
Für die monatlichen Ausgaben von 100 Studenten liegen nachfolgende klassierte Daten vor  
Klassen [200, 700) [700, 1200) [1200, 1700) [1700, 2200) [2200, 2700)  
relative Häufigkeit 0,1 0,3 0,3 0,2 0,1  
Berechnen Sie das arithmetische Mittel, die empirische Varianz und die Standardabweichung

```python
klassen = [200, 700, 1200, 1700, 2200, 2700]
rel_freq = [0.1, 0.3, 0.3, 0.2, 0.1]
klassenmittel = [(klassen[i] + klassen[i+1])/2 for i in range(len(rel_freq))]
mean = np.dot(klassenmittel, rel_freq)
var = np.dot((np.array(klassenmittel) - mean)**2, rel_freq)
std = np.sqrt(var)
print(f"Mittelwert: {mean}, Varianz: {var}, Std: {std}")
```
