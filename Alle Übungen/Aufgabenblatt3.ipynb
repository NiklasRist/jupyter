## Aufgabe 1
Gegeben sind die Datenpunkte A=(1|-2), B=(2|-1), C=(5|1) und D=(7|3).  
(a) Berechne die Fehlerquadratsumme für die lineare Funktion f1 mit  
$$
f_1(x) = 0{,}85\cdot x - 2{,}9
$$


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 5, 7])
y = np.array([-2, -1, 1, 3])
f1 = lambda x: 0.85 * x - 2.9
sse1 = np.sum((y - f1(x))**2)
print(f"Fehlerquadratsumme f1: {sse1:.4f}")
```

(b) Berechne die Fehlerquadratsumme für die lineare Funktion f2 mit  
$$
f_2(x) = 0{,}75 \cdot x - 2{,}5
$$

```python
f2 = lambda x: 0.75 * x - 2.5
sse2 = np.sum((y - f2(x))**2)
print(f"Fehlerquadratsumme f2: {sse2:.4f}")
```

(c) Die lineare Regressionsfunktion ist f3 mit  
$$f_3(x) = \frac{73}{91} x - \frac{251}{91}$$  
und Fehlerquadratsumme 0,1098.  
Rechnen Sie nach, dass ihr Funktionsgraph durch den Schwerpunkt S = (x|y) der Punktwolke verläuft.  

```python
f3 = lambda x: (73/91) * x - (251/91)
# Schwerpunkt berechnen
x_mean = np.mean(x)
y_mean = np.mean(y)
print(f"Schwerpunkt S: ({x_mean:.2f}, {y_mean:.2f})")

# Überprüfung, dass f3 durch den Schwerpunkt verläuft
f3_at_S = f3(x_mean)
print(f"f3(x_S) = {f3_at_S:.4f}")
print("Entspricht y_S?", np.isclose(f3_at_S, y_mean))
```

## Aufgabe 2
Der Zusammenhang zwischen der Absprunggeschwindigkeit und der Sprungweite soll  
untersucht werden. Es wird vermutet, dass die Sprungweite linear von der  
Absprunggeschwindigkeit abhängt.  
Es stehen folgende Messdaten zur Verfügung:  

| Absprunggeschwindigkeit (km/h) | 88,0 | 89,9 | 90,2 | 91,2 | 91,5 | 91,9 | 92,5 |  
|-------------------------------|:----:|:----:|:----:|:----:|:----:|:----:|:----:|  
| Sprungweite (m)               | 110,0| 112,5| 113,7| 115,8| 116,6| 118,7| 120,0|  

(a) Bestimmen Sie für diese Datenpaare eine Gleichung der linearen Regressionsfunktion.  

```python
# Messdaten
x = np.array([88.0, 89.9, 90.2, 91.2, 91.5, 91.9, 92.5])
y = np.array([110.0, 112.5, 113.7, 115.8, 116.6, 118.7, 120.0])

# Lineare Regression durchführen
m, b = np.polyfit(x, y, 1)
print(f"Regressionsgerade: y = {m:.4f}x + {b:.4f}")
```

(b) Interpretieren Sie den Wert der Steigung dieser Regressionsfunktion.  
Die Steigung gibt an, wie viele Meter zusätzlich an Sprungweite im Durchschnitt erwartet werden, wenn die Absprunggeschwindigkeit um 1 km/h steigt.


## Aufgabe 3
Gegeben sind Leistung und Verbrauch von sechs VW-Golf-Benzinmotoren  

| Leistung (kW)               | 55   | 74   | 77   | 85   | 110  | 150  |  
|-----------------------------|:----:|:----:|:----:|:----:|:----:|:----:|  
| Verbrauch (l/100km)         | 6,4  | 7,6  | 6,8  | 7,9  | 9,3  | 10,8 |  

(a) Stellen Sie die Daten in einem Streudiagramm dar  

```python

# Messdaten
leistung = np.array([55, 74, 77, 85, 110, 150])
verbrauch = np.array([6.4, 7.6, 6.8, 7.9, 9.3, 10.8])

# Streudiagramm erstellen
plt.scatter(leistung, verbrauch)
plt.xlabel('Leistung (kW)')
plt.ylabel('Verbrauch (l/100km)')
plt.title('Leistung vs. Verbrauch VW-Golf-Benzinmotoren')
plt.show()
```

(b) Bestimmen Sie die lineare Regressionsgerade und den Korrelationskoeffizienten 

```python
m, b = np.polyfit(leistung, verbrauch, 1)
print(f"Regressionsgerade: y = {m:.4f}x + {b:.4f}")

r = np.corrcoef(leistung, verbrauch)[0, 1]
print(f"Korrelationskoeffizient: r = {r:.4f}")
```

(c) Sie kaufen einen VW-Golf-Benzinmotor mit 130PS ≈ 130 * 0,74 kW = 96,2 kW. Mit   
welchem Verbrauch auf 100 km können Sie rechnen?  

```python
verbrauch_96_2 = m * 96.2 + b
print(f"Geschätzter Verbrauch für 96.2 kW: {verbrauch_96_2:.2f} l/100km")
```

```python
verbrauch_222 = m * 222 + b
print(f"Geschätzter Verbrauch für 222 kW: {verbrauch_222:.2f} l/100km")
```
