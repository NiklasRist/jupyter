1. (4P)
Gegeben seien 2 Merkmale X und Y, die an denselben Merkmalsträgern erhoben werden.
Das arithmetische Mittel des Merkmals X sei 𝑥̄ = 2 mit Standardabweichung 𝑠ₓ = 1
und das arithmetische Mittel des Merkmals Y sei 𝑦̄ = 4 mit Standardabweichung 𝑠ᵧ = 2.
Der Korrelationskoeffizient zwischen den Merkmalen sei r = 1.
Geben Sie die Regressionsgerade an und erstellen Sie ein Streudiagramm für X und Y.

```python
import numpy as np
import matplotlib.pyplot as plt

x_mean = 2
sx = 1
y_mean = 4
sy = 2
r = 1

b = r * (sy / sx)
a = y_mean - b * x_mean

print(f"Regressionsgerade: y = {a} + {b}x")

np.random.seed(0)
X = np.linspace(x_mean - 3*sx, x_mean + 3*sx, 10)
Y = a+b*X

plt.scatter(X, Y)
```

5. (4P)
X gibt die Augenzahl eines defekten Würfels beim einmaligen Würfeln an, mit folgender Wahrscheinlichkeitsfunktion:

k	        1	 2	    3	  4	    5	    6
P(X = k)	0.3	0.15	0	0.15	0.2 	0.2

Bestimmen Sie die Wahrscheinlichkeit, dass beim 49-maligen Würfeln die Augensumme ≥ 180 ist.

```python
import numpy as np
from scipy import stats

k= np.array([1,2,3,4,5,6])
p = np.array([0.3,	0.15,	0,	0.15,	0.2, 	0.2])

E_X = np.sum(k*p)
Var_X = np.sum((k-E_X)**2 * p)

n = 49

E_S = n*E_X
Var_S = n * Var_X
SD_S = np.sqrt(Var_S)

z = (180 - E_S) / SD_S
P = 1 - stats.norm.cdf(z)

print(f"Erwartungswert E[S] = {E_S:.3f}")
print(f"Standardabweichung SD[S] = {SD_S:.3f}")
print(f"Z-Wert = {z:.3f}")
print(f"P(S >= 180) = {P:.4f}")
```
