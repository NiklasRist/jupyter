Signifikanzniveau von 5% einigen.

ist.

101, 98, 96, 102,99, 101, 102, 103, 96, 97
Der LKW-Hersteller will nun wissen, ob er einen Preisnachlass fordern darf

```python
import numpy as np
from scipy import stats

data = np.array([101, 98, 96, 102,99, 101, 102, 103, 96, 97])

# 3️⃣ t-Test gegen µ0 = 100 (linksseitig)

n = len(data)
mean = np.mean(data) 
std = np.std(data, ddof=1) # Stichprobenstandardabweichung (empirsiche Varianz)
mu0 = 100

t_stat = (mean - mu0) / (std / np.sqrt(n))
df = n-1

p_wert = stats.t.cdf(t_stat, df=df) # CDF, weil linksseitig
t_crit = stats.t.ppf(0.05, df=df) # kritischer t-Wert bei α=5% linksseitig

print(t)
print(p_wert)
print(t_crit)
```
