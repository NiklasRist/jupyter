## Aufgabe 1

Von den 30 Einzelhändlern in einer Kleinstadt sind 10 für und 20 gegen eine Verlängerung der Ladenöffnungszeiten. Im Rahmen einer Umfrage werden 6 zufällig ausgewählte Einzelhändler nach ihrer Meinung befragt.

(a) Wie groß ist die Wahrscheinlichkeit, dass sich genau zwei der befragten Einzelhändler für eine Verlängerung der Ladenöffnungszeiten aussprechen?

(b) Wie groß ist die Wahrscheinlichkeit, dass sich mindestens die Hälfte der befragten Händler für eine Verlängerung der Ladenöffnungszeiten ausspricht?

(c) Wie viele Einzelhändler sind in der Umfrage zu erwarten, die sich für eine Verlängerung der Ladenöffnungszeit aussprechen?

```python
from scipy.stats import hypergeom
n = 6
N = 30
dafür = 10
p = dafür/N

print(f"a) P(2) = {hypergeom.pmf(2, N, dafür, n):.4f}")
print(f"b) P(X >= 3) = {hypergeom.sf(2, N, dafür, n ):.4f}")
print(f"c) E(X) = {hypergeom.mean(N, dafür, n):.4f}")
```

## Aufgabe 3

Ein Angestellter verlegt seinen Arbeitsplatz an den heimischen Schreibtisch und wechselt dabei den Provider. Bei seinen vielen Einwahlen kommen pro Monat im Mittel 3 Fehlversuche vor.

(a) Geben Sie ein sinnvolles Verteilungsmodell für die Anzahl der Fehlversuche pro Monat an (Typ, Parameter).

(b) Wie groß sind Erwartungswert und Varianz der Anzahl der Fehlversuche pro Monat?

(c) Wie groß ist die Wahrscheinlichkeit, dass er einen Monat ohne Fehlversuche erlebt?

```python
from scipy.stats import poisson
k = 0
print(f"a) P(0) = {poisson.pmf(k, 3)}")
print(f"b) E(X) = {poisson.mean(3)}")
print(f"b) Var(X) = {poisson.var(3)}")
print(f"c) P(0) = {poisson.pmf(0, 3)}")
```
