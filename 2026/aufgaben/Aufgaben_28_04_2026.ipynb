## Aufgabe 1

Gegeben sei eine diskrete Zufallsvariable X.

Für die möglichen Realisationen xi von X seien folgende Wahrscheinlichkeiten bekannt:

| xi | 0 | 1 | 2 | 3 |
|----|---|---|---|---|
| f(xi) | 0.1 | 0.5 − c | 0.5 − c | c |

a) Bestimmen Sie die Konstante c.

b) Bestimmen Sie die zugehörige Wahrscheinlichkeits- und die zugehörige Verteilungsfunktion

c) Bestimmen Sie den Erwartungswert und die Varianz von X.

d) Bestimmen Sie die Wahrscheinlichkeiten:
   - P(X ≤ 2,5)
   - P(1,5 < X ≤ 3)
   - P(X > 1)

## Aufgabe 2

Die folgende Tabelle gibt die Wahrscheinlichkeitsverteilung der Anzahl der Personen pro Fahrzeug an:

| Anzahl der Personen | 1 | 2 | 3 | 4 | 5 | 6 |
|---------------------|---|---|---|---|---|---|
| Wahrscheinlichkeit  | 0,05 | 0,43 | 0,27 | 0,12 | 0,09 | 0,04 |

Die folgenden alternativen Möglichkeiten zur Preisgestaltung werden in Betracht gezogen:

- 0,50 € pro Fahrzeuginsasse
- 0,50 € sowohl für das Fahrzeug als auch den Fahrer und 0,35 € für jeden weiteren Insassen

Berechnen Sie die erwarteten Einnahmen pro Fahrzeug für jede der beiden Alternativen.

## Aufgabe 3

Gegeben sei die Funktion:

f(x) = {
  cx − 0,5   für 1 ≤ x ≤ 3
  0          sonst
}

a) Bestimmen Sie die Konstante c so, dass f(x) eine Dichtefunktion ist.
b) Bestimmen Sie den Erwartungswert und die Varianz von X.
c) Bestimmen Sie folgende Wahrscheinlichkeiten:
   - P(X = 2)
   - P(X > 2,5)
   - F(2,5)
d) Bestimmen Sie ein Intervall, sodass sich links 6,25% und rechts 19% Wahrscheinlichkeitsmasse befinden.
