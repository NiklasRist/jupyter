## 1.

Der Zusammenhang zwischen der alter und der Sprungweite soll untersucht werden.  
Es wird vermutet, dass die Sprungweite linear von der alter abhängt.  
Es stehen folgende Messdaten zur Verfügung:

| Absprunggeschwindigkeit (km/h) | 88,0 | 89,9 | 90,2 | 91,2 | 91,5 | 91,9 | 92,5 |
|-------------------------------|------|------|------|------|------|------|------|
| Sprungweite (m)               | 110,0 | 112,5 | 113,7 | 115,8 | 116,6 | 118,7 | 120 |

(a) Bestimmen Sie für diese Datenpaare eine Gleichung der linearen Regressionsfunktion.  
(b) Interpretieren Sie den Wert der Steigung dieser Regressionsfunktion.


```python
import math
absprunggeschwindigkeit = [88, 89.9, 90.2, 91.2, 91.5, 91.9, 92.5]
sprungweite = [110, 112.5, 113.7, 115.8, 116.6, 118.7, 120]

n=len(absprunggeschwindigkeit)
x_mittel = sum(absprunggeschwindigkeit)/n
y_mittel = sum(sprungweite)/n

s_x_2 = sum((x-x_mittel)**2 for x in absprunggeschwindigkeit)/(n-1)
s_y_2 = sum((y-y_mittel)**2 for y in sprungweite)/(n-1)

s_x = math.sqrt(s_x_2)
s_y = math.sqrt(s_y_2)

s_xy = sum((x-x_mittel)*(y-y_mittel) for x, y in zip(absprunggeschwindigkeit, sprungweite))/(n-1)
s_x_s_y = s_x * s_y

r_xy = s_xy/s_x_s_y

k = r_xy*(s_y/s_x)
d = y_mittel-k*x_mittel

print(f"a) y={k}x+{d}")
print(f"b) Die Wert der Steigung k = {k} bedeutet, dass eine Erhöhung der Absprunggeschwindigkeit um 1 km/h in einer Zunahme der Sprungweite um {k} m resultiert.")
```

## 2.

Von 4 Kfz sind das Alter und die Bremswege bei einer Vollbremsung von 100 km/h zum Stillstand gegeben:

| Alter (Jahre) | 4 | 7 | 11 | 2 |
|---------------|---|---|----|---|
| Bremsweg (m)  | 50 | 80 | 70 | 45 |

(a) Zeichnen Sie ein Streudiagramm.  
(b) Bestimmen Sie die Gleichung der Regressionsgeraden und den Korrelationskoeffizienten.  
(c) Extrapolieren Sie den erwarteten mittleren Bremsweg für ein 13 Jahre altes Fahrzeug.


```python
import numpy as np 
import matplotlib.pyplot as plt

alter = np.array([4, 7, 11, 2])
bremsweg = np.array([50, 80, 70, 45])

# a

plt.scatter(alter, bremsweg)
plt.xlabel("Alter in Jahren")
plt.ylabel("Bremsweg in m")
plt.grid(True)
plt.show()

# b

n=len(alter)
x_mittel = sum(alter)/n
y_mittel = sum(bremsweg)/n

s_x_2 = sum((x-x_mittel)**2 for x in alter)/(n-1)
s_y_2 = sum((y-y_mittel)**2 for y in bremsweg)/(n-1)

s_x = math.sqrt(s_x_2)
s_y = math.sqrt(s_y_2)

s_xy = sum((x-x_mittel)*(y-y_mittel) for x, y in zip(alter, bremsweg))/(n-1)
s_x_s_y = s_x * s_y

r_xy = s_xy/s_x_s_y

k = r_xy*(s_y/s_x)
d = y_mittel-k*x_mittel

print(f"y={k}x+{d}")
print(f"Korrelationskoeffizient r_xy = {r_xy}")

# c

print(f"f(13) = {k*13+d}")
```
