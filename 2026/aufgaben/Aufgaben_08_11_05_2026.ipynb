## Aufgabe 1

Eine Serienproduktion von Transistoren hat einen gleichbleibenden Ausschuss-Anteil von $p = 3\%$.

Wie groß ist die Wahrscheinlichkeit, dass in einer Stichprobe von $n = 50$ entnommenen Einheiten

(a) keine

(b) genau zwei

(c) höchstens zwei

fehlerhafte Bauteile sind?

Wie groß sind Erwartungswert und Varianz der Anzahl $X$ der fehlerhaften Bauteile in der Stichprobe?

```python
from scipy.stats import binom

print(f"a) P(0) = {binom.pmf(0, 50, 0.03):.4f}")
print(f"b) P(2) = {binom.pmf(2, 50, 0.03):.4f}")
print(f"c) P(X <= 2) = {binom.cdf(2, 50, 0.03):.4f}")
print(f"µ = {binom.mean(50, 0.03):.4f}")
print(f"σ^2 = {binom.var(50, 0.03):.4f}")
```

## Aufgabe 2

Ein Computerhersteller will eine neue Bestückungsmaschine für Platinen kaufen. Die Ausschussrate soll höchstens $5\%$ sein. Zur Kontrolle wird ein Probelauf mit 20 Platinen durchgeführt. Sind mehr als $k$ Platinen fehlerhaft bestückt, so muss die Produktion gestoppt und kostenfrei nachgebessert werden.

(a) Mit welcher Wahrscheinlichkeit erhält man bei einer tatsächlichen Ausschussrate von $5\%$ höchstens 3 fehlerhafte Platinen?

(b) Wie muss die Zahl $k$ gewählt werden, damit die Wahrscheinlichkeit für einen Produktionsstopp trotz ausreichender Ausschussrate kleiner als $10\%$ ist?

```python
print(f"a) P(X <= 3) = {binom.cdf(3, 20, 0.05):.4f}")
k = int(binom.isf(0.1, 20, 0.05))
print(f"b) k = {k}")
print(f"Kontrolle Wahrscheinlichkeit  für falschen Produktionsstopp: P(X > {k}) = {binom.sf(k, 20, 0.05):.4f}")
```

## Aufgabe 3

Wie oft muss man mindestens würfeln, um mit einer Wahrscheinlichkeit von mindestens $90\%$ mindestens eine Sechs zu bekommen?

```python
import math

# brute force
n = 1

while binom.sf(0, n, 1/6) < 0.9:
    n += 1

print(f"n = {n}")
print(f"Wahrscheinlichkeit = {binom.sf(0, n, 1/6):.4f}")

# elegant
print("---------------------------")
print("P(X >= 1) = 1 - P(X = 0)")
print("1-(5/6)^n = 0,9 ")
print("(5/6)^n = 0,1 ")
print(f"n = {math.ceil(math.log(0.1) / math.log(5/6))}")
```
