## Aufgabe 1

Gegeben ist die quadratische Gleichung $x^2 + px + q = 0$, wobei $p$ und $q$ Zufallszahlen im Intervall $[0,4]$ sind.

Gesucht ist die Wahrscheinlichkeit, dass die quadratische Gleichung mindestens eine reelle Lösung hat.

**Hinweis:** Skizzieren Sie im Bereich mit $0 \leq q \leq 4$ und $0 \leq p \leq 4$ den Graph der Funktion
$$q = \frac{p^2}{4}$$
und markieren Sie den Bereich mit
$$q \leq \frac{p^2}{4}.$$

```python
import numpy as np
import matplotlib.pyplot as plt


p = np.linspace(0, 4, 400)
q = p**2 / 4


plt.plot(p, q)


plt.fill_between(p, q)


plt.xlim(0, 4) 
plt.ylim(0, 4)


plt.xlabel('p')
plt.ylabel('q')


plt.grid()

plt.show()


# Integral berechnen

obere_grenze = 4
untere_grenze = 0

fläche = (1/4) * (obere_grenze**3 / 3 - untere_grenze**3 / 3)
wahrscheinlichkeit = fläche / 16

print(f"Fläche = {fläche}")
print(f"Die Wahrscheinlichkeit, dass die quadratische Gleichung mindestens eine reelle Lösung hat, ist {wahrscheinlichkeit}, da die Formel im Hinweis die umgeformte Diskriminante ist.")
```

## Aufgabe 2

In einer Gruppe von 900 Personen haben sich 600 prophylaktisch gegen Grippe impfen lassen. Nach einer bestimmten Zeit wurde jedes Gruppenmitglied danach befragt, wer an einer Grippe erkrankte. Die Ergebnisse werden in einer 4-Feldtafel dargestellt.

| Gruppe | B erkrankt | B gesund | Summe |
|---|---:|---:|---:|
| A geimpft | 60 | 540 | 600 |
| A ungeimpft | 120 | 180 | 300 |
| **Summe** | **180** | **720** | **900** |

Berechnen Sie:

- $P(A)$
- $P(B)$
- $P(A \cap B)$
- $P(B\mid A)$
- $P(A\mid B)$
- $P(¬A \cap B)$
- $P(B\mid ¬A)$

```python
a = 600
a_nicht = 300
b = 180
b_nicht = 720
a_b = 60
a_b_nicht = 540
a_nicht_b = 120
a_nicht_b_nicht = 180
gesamt = 900

p_a = a/gesamt
p_a_nicht = 1-p_a
p_b = b/gesamt
p_a_b = a_b/gesamt
p_a_bed_b = p_a_b/p_b
p_b_bed_a = p_a_b/p_a
p_a_nicht_b = a_nicht_b/gesamt
p_b_bed_a_nicht = p_a_nicht_b/p_a_nicht

print(f"P(A) = {p_a}")
print(f"P(B) = {p_b}")
print(f"P(A ∩ B) = {p_a_b}")
print(f"P(B ∣ A) = {p_b_bed_a}")
print(f"P(A ∣ B) = {p_a_bed_b}")
print(f"P(¬A ∩ B) = {p_a_nicht_b}")
print(f"P(B ∣ ¬A) = {p_b_bed_a_nicht}")
```

## Aufgabe 3

Bei einer Sportveranstaltung wird ein Dopingtest durchgeführt. Wenn ein Sportler gedopt hat, dann fällt der Test zu 99% positiv aus. Hat ein Sportler aber nicht gedopt, zeigt der Test trotzdem zu 5% ein positives Ergebnis an. Aus Erfahrung weiß man, dass 20% der Sportler gedopt sind.

**(a)** Bestimmen Sie die Wahrscheinlichkeit, dass eine Dopingprobe positiv ausfällt.

**(b)** Bestimmen Sie die Wahrscheinlichkeit, dass der Test negativ ausfällt, obwohl der Sportler gedopt hat.

**(c)** Berechnen Sie die Wahrscheinlichkeit, dass ein Sportler gedopt hat, wenn seine Dopingprobe negativ war.

```python
p_positiv_bed_gedopt = 0.99
p_positiv_bed_nicht_gedopt = 0.05
p_gedopt = 0.2
p_nicht_gedopt = 1-p_gedopt

# a
p_positiv_gedopt = p_positiv_bed_gedopt*p_gedopt
p_positiv_nicht_gedopt = p_positiv_bed_nicht_gedopt *p_nicht_gedopt
p_positiv = p_positiv_gedopt + p_positiv_nicht_gedopt

# b
p_negativ_bed_gedopt = 1-p_positiv_bed_gedopt

# c
p_negativ = 1-p_positiv
p_gedopt_bed_negativ = (p_negativ_bed_gedopt*p_gedopt)/p_negativ

print(f"a) Die Wahrscheinlichkeit, dass eine Dopingprobe positiv ausfällt beträgt {p_positiv}")
print(f"b) Die Wahrscheinlichkeit, dass der Test negativ ausfällt, obwohl der Sportler gedopt hat, beträgt: {p_negativ_bed_gedopt}")
print(f"c) Die Wahrscheinlichkeit, dass ein Sportler gedopt hat, wenn seine Dopingprobe negativ war, beträgt: {p_gedopt_bed_negativ}")
```
