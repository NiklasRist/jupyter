## 1. Dichtefunktion und Quantilsfunktion

Es sei $X$ eine Zufallsvariable mit Dichtefunktion

$$
f(t) =
\begin{cases}
0{,}5t - 0{,}5 & \text{für } 1 \le t \le 3 \\
0 & \text{sonst}
\end{cases}
$$

Erstellen Sie bitte zu dieser Dichtefunktion ein Python-Programm, welches für einen Input

$p \in [0;1]$

als Output den Wert

$a \in [1;3]$

liefert mit

$$
\int_{-\infty}^{a} f(t)\,dt = p
$$

Vergleichbar zu den `ppf`-Methoden von Python.

```python
F = lambda t: 0.25 * t**2 - 0.5 * t - (0.25 * 1**2 - 0.5 * 1)

def ppfBruteForce(p):
    if p < 0 or p > 1:
        return None
    p = round(p, 3)
    for i in range(1000, 3001):
        i /= 1000
        if abs(F(i) - p) < 1e-3:
            return i
    return None

print(ppfBruteForce(0.67))

```

## 2. Erwarteter Gewinn einer Abteilung

Eine Abteilung eines Maschinenbauunternehmens stellt Großanlagen eines bestimmten Typs her.

Die Wahrscheinlichkeiten, dass im Geschäftsjahr eine bestimmte Anzahl von Anlagen abgesetzt werden kann, haben folgende Werte:

| Anlagenzahl | 0 | 1 | 2 | 3 | 4 | 5 |
|---:|---:|---:|---:|---:|---:|---:|
| Wahrscheinlichkeit | 0,05 | 0,15 | 0,25 | 0,30 | 0,15 | 0,10 |

Die Fixkosten der Abteilung belaufen sich im Geschäftsjahr auf **1 Mio. EUR**.

Je gebauter Anlage entstehen variable Kosten in Höhe von **500.000 EUR**.

Der Erlös pro abgesetzter Anlage beträgt **1 Mio. EUR**.

Geben Sie den Gewinn als Funktion von Kosten, Erlös und Fixkosten an.

Berechnen Sie damit den erwarteten Gewinn der Abteilung.


```python
gewinn = lambda x: 1000000 * x - 500000 * x - 1000000

augenzahl = [0, 1, 2, 3, 4, 5]
wahrscheinlichkeit = [0.05, 0.15, 0.25, 0.30, 0.15, 0.10]

µ = sum(gewinn(augenzahl[i]) * wahrscheinlichkeit[i] for i in range(len(augenzahl)))

print(f"Erwarteter Gewinn: {µ} EUR")
```

## 3. Erwartete Kosten pro Tag

Bei einer Produktion können pro Tag folgende Anzahl an Störfällen mit den entsprechenden Wahrscheinlichkeiten auftreten:

| Anzahl | 0 | 1 | 2 | 3 |
|--------|---|---|---|---|
| p      | 0,60 | 0,12 | 0,20 | 0,08 |

Die Kosten $K(x)$ in **tsd. Euro** für die Behebung der einzelnen Störfälle können durch folgende Funktion ermittelt werden:

$$
K(x) = 6 - \frac{6}{1 + x}
$$

Was sind die zu erwartenden Kosten pro Tag?

```python
anzahl = [0, 1, 2, 3]
wahrscheinlichkeit = [0.60, 0.12, 0.20, 0.08]

kosten = lambda x: 6 - 6 / (1 + x)
µ = sum(kosten(anzahl[i]) * wahrscheinlichkeit[i] for i in range(len(anzahl)))

print(f"Erwartete Kosten: {µ} tsd. EUR")
```
