## Aufgabe 1

Gegeben ist eine Zufallsvariable \(X\) mit Dichtefunktion:

\[
f(t) =
\begin{cases}
0{,}5t - 0{,}5 & \text{für } 1 \le t \le 3 \\
0 & \text{sonst}
\end{cases}
\]

Erstellen Sie ein Python-Programm, das für einen Input \(p \in [0,1]\) den Wert \(a \in [1,3]\) liefert, sodass gilt:

\[
\int_{-\infty}^{a} f(t)\,dt = p
\]

(vergleichbar mit der ppf-Methode)


## Aufgabe 2

Eine Abteilung eines Maschinenbauunternehmens stellt Großanlagen her.

Gegeben:

| Anlagenzahl | 0 | 1 | 2 | 3 | 4 | 5 |
|------------|---|---|---|---|---|---|
| Wahrscheinlichkeit | 0,05 | 0,15 | 0,25 | 0,3 | 0,15 | 0,1 |

Weitere Daten:
- Fixkosten: 1 Mio. EUR
- Variable Kosten pro Anlage: 500.000 EUR
- Erlös pro Anlage: 1 Mio. EUR

Aufgaben:
1. Geben Sie den Gewinn als Funktion an.
2. Berechnen Sie den erwarteten Gewinn.


## Aufgabe 3
