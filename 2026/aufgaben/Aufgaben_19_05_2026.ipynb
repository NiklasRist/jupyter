1. Sie stehen im Supermarkt an der Kasse in der Warteschlange. Sie wissen, dass die durchschnittliche
Wartezeit 4 Minuten beträgt. Gehen Sie davon aus, dass die Wartezeit W (in Minuten)
exponentialverteilt ist.
Mit welcher Wahrscheinlichkeit warten Sie

(a) genau 3 Minuten?

(b) länger als 4 Minuten?

(c) zwischen 2 und 7 Minuten?

(d) weitere 2 Minuten, nachdem Sie schon 3 Minuten gewartet haben

Wie groß ist die Standardabweichung der Wartezeit?


```
from scipy.stats import expon

scale = 4
print(f"a) P(X = 3) = 0")
print(f"b) P(X > 4) = 1 - P(X <= 4) = {1 - expon.cdf(4, scale=scale)}")
print(f"c) P(2 <= X <= 7) = P(X <= 7) - P(X < 2) = {expon.cdf(7, scale=scale) - expon.cdf(2, scale=scale)}")
print(f"d) P(X >= 2 | X = 3) = P(X >= 2) = 1 - P(X < 2) = 1 - P(X <= 2) = {1 - expon.cdf(2, scale=scale)}")
print(f"e) Standardabweichung = {expon.std(scale=scale)}")
```

2. Eine Zufallsvariable X sei exponentialverteilt, d.h. die Dichtefunktion lautet:
f(t) = λe^−(λt) für t ≥ 0.
Berechnen Sie den Median Q0,5.

$$
\begin{aligned}
F(x_{med}) &= 0.5 \\
1 - e^{-\lambda \cdot x_{med}} &= 0.5 \\
e^{-\lambda \cdot x_{med}} &= 0.5 \\
-\lambda \cdot x_{med} &= \ln(0.5) \\
-\lambda \cdot x_{med} &= -\ln(2) \\
x_{med} &= \frac{\ln(2)}{\lambda}
\end{aligned}
$$



3. Ein Unternehmen produziert Vitamin-C-Tabletten, die jeweils 100 mg Vitamin-C enthalten
sollen. Produktionsbedingt ist mit einer Standardabweichung von 5 mg zu rechnen.
Ausgeliefert werden Tabletten mit einem Vitamin-C-Gehalt der zwischen 95 mg und 110
mg liegt. Berechnen Sie die Wahrscheinlichkeit, dass bei einer zufälligen Entnahme einer
Tablette diese im für die Auslieferung erlaubten Bereich liegt.

```
from scipy.stats import norm

µ = 100
σ = 5
untere_grenze = 95
obere_grenze = 110

print(f"a) P(95 <= X <= 110) = {norm.cdf(obere_grenze, µ, σ) - norm.cdf(untere_grenze, µ, σ)}")
```
