## 2.2 Defining Column Constraints and Data Types

```python
# ==========================================
# STEP 2: DEFINE CONSTRAINTS FOR EACH COLUMN
# ==========================================

CONSTRAINTS = {
    'DATE_TIME': {
        'description': 'Timestamp of measurement',
        'expected_type': 'datetime',
        'format_variants': ['%Y-%m-%d %H:%M:%S'],
        'range': ('2020-05-15 00:00:00', '2020-06-17 23:45:00'),
        'nullable': False
    },
    'PLANT_ID': {
        'description': 'Solar plant identifier',
        'expected_type': 'integer',
        'allowed_values': [4136001],
        'nullable': False
    },
    'SOURCE_KEY': {
        'description': 'Inverter/Generator identifier',
        'expected_type': 'string',
        'nullable': False
    },
    'DC_POWER': {
        'description': 'Direct Current power (kW)',
        'expected_type': 'float',
        'min': 0.0,
        'max': 14000.000000,
        'nullable': False,
        'decimal_separator': '.'
    },
    'AC_POWER': {
        'description': 'Alternating Current power (kW)',
        'expected_type': 'float',
        'min': 0.0,
        'max': 1300.000000,
        'nullable': False,
        'decimal_separator': '.'
    },
    'DAILY_YIELD': {
        'description': 'Daily energy production (kWh)',
        'expected_type': 'float',
        'min': 0.0,
        'max': 9250.000000,
        'nullable': False,
        'decimal_separator': '.'
    },
    'TOTAL_YIELD': {
        'description': 'Cumulative energy production (kWh)',
        'expected_type': 'float',
        'min': 0.0,
        'max': 1.215279e+09,
        'nullable': False,
        'decimal_separator': '.'
    }
}

print("=== CONSTRAINTS DEFINED ===")
for col, rules in CONSTRAINTS.items():
    print(f"\n{col}: {rules['description']}")
    for rule, value in rules.items():
        if rule != 'description':
            print(f"  - {rule}: {value}")
```

## 3. Data Cleaning
#### 3.1. Check for missing values

```python
# Check for missing values
missing_values = plant2_gen.isnull().sum()
print("Missing values per column:")
print(missing_values)
print("\nPercentage missing:")
print((missing_values / len(plant2_gen)) * 100)
```

#### 3.2. Check for duplicate values


```python
# Check for duplicate rows
duplicate_rows = plant2_gen.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_rows}")

if duplicate_rows > 0:
    print("\nDuplicate rows (first 5):")
    print(plant2_gen[plant2_gen.duplicated(keep=False)].head())
```

#### 3.3. Check for outliers


```python
# Check for outliers using IQR method
import numpy as np

numeric_columns = ['DC_POWER', 'AC_POWER', 'DAILY_YIELD', 'TOTAL_YIELD']

for col in numeric_columns:
    Q1 = plant2_gen[col].quantile(0.25)
    Q3 = plant2_gen[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = plant2_gen[(plant2_gen[col] < lower_bound) | (plant2_gen[col] > upper_bound)]
    print(f"\n{col}:")
    print(f"  Lower bound: {lower_bound:.2f}")
    print(f"  Upper bound: {upper_bound:.2f}")
    print(f"  Number of outliers: {len(outliers)}")
    if len(outliers) > 0:
        print(f"  Outlier range: {outliers[col].min():.2f} to {outliers[col].max():.2f}")
```

#### 3.4. Check for consistency in data


```python
# Check for data consistency
print("=== DATA CONSISTENCY CHECKS ===")

# Check if AC_POWER is always <= DC_POWER (should be true due to conversion losses)
ac_gt_dc = plant2_gen[plant2_gen['AC_POWER'] > plant2_gen['DC_POWER']]
print(f"\nAC_POWER > DC_POWER: {len(ac_gt_dc)} records")
if len(ac_gt_dc) > 0:
    print("Sample records where AC > DC:")
    print(ac_gt_dc[['DATE_TIME', 'DC_POWER', 'AC_POWER']].head())

# Check if DAILY_YIELD resets to 0 at midnight
plant2_gen['DATE_TIME_DT'] = pd.to_datetime(plant2_gen['DATE_TIME'], format='%Y-%m-%d %H:%M:%S')
plant2_gen['HOUR'] = plant2_gen['DATE_TIME_DT'].dt.hour
midnight_resets = plant2_gen[(plant2_gen['HOUR'] == 0) & (plant2_gen['DAILY_YIELD'] > 0)]
print(f"\nMidnight records with DAILY_YIELD > 0: {len(midnight_resets)}")

# Check if TOTAL_YIELD is monotonically increasing within each SOURCE_KEY
print(f"\nChecking TOTAL_YIELD monotonicity per SOURCE_KEY...")
non_monotonic = []
for source_key in plant2_gen['SOURCE_KEY'].unique():
    source_data = plant2_gen[plant2_gen['SOURCE_KEY'] == source_key].sort_values('DATE_TIME_DT')
    if not source_data['TOTAL_YIELD'].is_monotonic_increasing:
        non_monotonic.append(source_key)

print(f"SOURCE_KEYs with non-monotonic TOTAL_YIELD: {len(non_monotonic)}")
if len(non_monotonic) > 0:
    print(f"Problematic SOURCE_KEYs: {non_monotonic[:5]}...")

# Clean up temporary columns
plant2_gen.drop(['DATE_TIME_DT', 'HOUR'], axis=1, inplace=True, errors='ignore')
```

#### 3.5. Check for data type consistency

```python
# Check data type consistency
print("=== DATA TYPE CONSISTENCY CHECKS ===")

# Expected types based on constraints
expected_types = {
    'DATE_TIME': 'object',  # Will be converted to datetime
    'PLANT_ID': 'int64',
    'SOURCE_KEY': 'object',
    'DC_POWER': 'float64',
    'AC_POWER': 'float64',
    'DAILY_YIELD': 'float64',
    'TOTAL_YIELD': 'float64'
}

print("\nCurrent data types:")
print(plant2_gen.dtypes)

print("\nData type validation:")
type_issues = []
for col, expected_type in expected_types.items():
    actual_type = str(plant2_gen[col].dtype)
    if actual_type != expected_type:
        type_issues.append(f"{col}: expected {expected_type}, got {actual_type}")
        print(f"  ❌ {col}: expected {expected_type}, got {actual_type}")
    else:
        print(f"  ✅ {col}: {actual_type}")

if type_issues:
    print(f"\n⚠️  Found {len(type_issues)} data type issues")
else:
    print("\n✅ All data types are as expected")

# Test DATE_TIME conversion
print("\nTesting DATE_TIME format conversion...")
try:
    test_dates = pd.to_datetime(plant2_gen['DATE_TIME'].head(10), format='%Y-%m-%d %H:%M:%S')
    print("✅ DATE_TIME format is valid")
    print(f"Sample converted dates: {test_dates.tolist()[:3]}")
except Exception as e:
    print(f"❌ DATE_TIME format error: {e}")
```

#### 3.6. Check for plausibility of data


```python
# Check data plausibility
print("=== DATA PLAUSIBILITY CHECKS ===")

# Check against defined constraints
constraint_violations = []

# 1. Check PLANT_ID values
invalid_plant_ids = plant2_gen[~plant2_gen['PLANT_ID'].isin(CONSTRAINTS['PLANT_ID']['allowed_values'])]
if len(invalid_plant_ids) > 0:
    constraint_violations.append(f"PLANT_ID: {len(invalid_plant_ids)} invalid values")

# 2. Check numeric ranges
for col in ['DC_POWER', 'AC_POWER', 'DAILY_YIELD', 'TOTAL_YIELD']:
    min_val = CONSTRAINTS[col]['min']
    max_val = CONSTRAINTS[col]['max']

    below_min = plant2_gen[plant2_gen[col] < min_val]
    above_max = plant2_gen[plant2_gen[col] > max_val]

    if len(below_min) > 0:
        constraint_violations.append(f"{col}: {len(below_min)} values below minimum ({min_val})")

    if len(above_max) > 0:
        constraint_violations.append(f"{col}: {len(above_max)} values above maximum ({max_val})")

# 3. Check date range
try:
    dates = pd.to_datetime(plant2_gen['DATE_TIME'], format='%Y-%m-%d %H:%M:%S')
    min_date = pd.to_datetime(CONSTRAINTS['DATE_TIME']['range'][0], format='%Y-%m-%d %H:%M:%S')
    max_date = pd.to_datetime(CONSTRAINTS['DATE_TIME']['range'][1], format='%Y-%m-%d %H:%M:%S')

    invalid_dates = dates[(dates < min_date) | (dates > max_date)]
    if len(invalid_dates) > 0:
        constraint_violations.append(f"DATE_TIME: {len(invalid_dates)} values outside expected range")
except Exception as e:
    constraint_violations.append(f"DATE_TIME: Format conversion error - {e}")

# 4. Business logic checks
# Check for negative power values (shouldn't exist)
negative_dc = plant2_gen[plant2_gen['DC_POWER'] < 0]
negative_ac = plant2_gen[plant2_gen['AC_POWER'] < 0]

if len(negative_dc) > 0:
    constraint_violations.append(f"DC_POWER: {len(negative_dc)} negative values")

if len(negative_ac) > 0:
    constraint_violations.append(f"AC_POWER: {len(negative_ac)} negative values")

# 5. Check for unusual ratios
# AC/DC ratio should typically be between 0.9-0.98 for modern inverters
dc_positive = plant2_gen[plant2_gen['DC_POWER'] > 0]
if len(dc_positive) > 0:
    ratios = dc_positive['AC_POWER'] / dc_positive['DC_POWER']
    unusual_ratios = ratios[(ratios < 0.8) | (ratios > 1.0)]
    if len(unusual_ratios) > 0:
        constraint_violations.append(f"AC/DC ratio: {len(unusual_ratios)} unusual values (<0.8 or >1.0)")

print(f"\nFound {len(constraint_violations)} constraint violations:")
for violation in constraint_violations:
    print(f"  ⚠️  {violation}")

if len(constraint_violations) == 0:
    print("\n✅ All constraint checks passed!")
else:
    print(f"\n❌ Data quality issues detected. Review required.")
```
