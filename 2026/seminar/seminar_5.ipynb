```python
gesamt = 674
männlich = 293
männlich_rauchen = 82
weiblich = gesamt - männlich
weiblich_nicht_rauchen = 250
weiblich_rauchen = weiblich - weiblich_nicht_rauchen

p_männlich = männlich/gesamt
p_männlich_rauchen = männlich_rauchen/gesamt
p_männlich_bed_rauchen = p_männlich*(p_männlich_rauchen/p_männlich)

print(f"a) Eine zufällig ausgewählte Person ist mit {(100*weiblich_nicht_rauchen)/gesamt} % eine Nichtraucherin.")
print(f"b) Die Schülerin im Aufenthaltsraum ist zu {weiblich_nicht_rauchen/weiblich*100} % Nichtraucherin.")
print(f"c) Die Ereignisse männlich und Rauchen sind voneinander abhängig: {round(p_männlich_rauchen,4) == round(p_männlich_bed_rauchen,4)}")
```
