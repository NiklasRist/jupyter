7. 6% der europäischen Bevölkerung haben Blutgruppe 0-, welche für alle Empfänger verträglich
ist

(a) Berechnen Sie die Anzahl der Spender, die benötigt werden, um mit mindestens
95%iger Wahrscheinlichkeit mindestens einen Spender mit Blutgruppe 0- zu haben.

(b) Bestimmen Sie die Anzahl der Spender, die benötigt werden, um mit mindestens
90%iger Wahrscheinlichkeit mindestens zwei Spender mit Blutgruppe 0- zu haben.

(c) Bestimmen Sie die Anzahl der Spender, die benötigt werden, um mit mindestens
85%iger Wahrscheinlichkeit mindestens vier Spender mit Blutgruppe 0- zu haben.

```python
from scipy.stats import binom

p_gruppe_0_minus = 0.06  # 6% der Bevölkerung haben Blutgruppe 0-

def finde_anzahl_spender(target_wahrscheinlichkeit, p_gruppe_0_minus, min_anzahl):
    """
    Findet die minimale Anzahl an Spendern, um mit mindestens target_wahrscheinlichkeit
    mindestens min_anzahl Spender mit Blutgruppe 0- zu haben.
    """
    for n in range(1, 1000):
        # Wahrscheinlichkeit, mindestens min_anzahl Spender mit Blutgruppe 0- zu haben
        # = 1 - Wahrscheinlichkeit, weniger als min_anzahl Spender zu haben
        wahrscheinlichkeit = 1 - binom.cdf(min_anzahl - 1, n, p_gruppe_0_minus)

        if wahrscheinlichkeit >= target_wahrscheinlichkeit:
            print(f"{wahrscheinlichkeit} {target_wahrscheinlichkeit}")
            return n

    return None

print(f"a) Anzahl der Spender: {finde_anzahl_spender(0.95, p_gruppe_0_minus, 1)}")
print(f"b) Anzahl der Spender: {finde_anzahl_spender(0.90, p_gruppe_0_minus, 2)}")
print(f"c) Anzahl der Spender: {finde_anzahl_spender(0.85, p_gruppe_0_minus, 4)}")
```
