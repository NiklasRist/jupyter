1. Die durchschnittliche Lebensdauer (Nutzungsdauer) einer Maschine sei 10 Jahre. Die
Wahrscheinlichkeit, dass die Maschine während der Laufzeit kaputt geht, sei exponentialverteilt.
Wie hoch ist die Wahrscheinlichkeit, dass die Maschine innerhalb der ersten 5 Jahre kaputt
geht?

```python
from scipy.stats import expon
lebensdauer = 10
lambda_ = 1 / lebensdauer

expon.cdf(5*lambda_)
```

3. Eine Glühbirne in einer Notbeleuchtung ist ununterbrochen in Betrieb bis sie ausfällt. Die
Zufallsvariable X, durch die die Lebensdauer (in Stunden) von Glühbirnen modelliert
wird, sei exponentialverteilt mit Parameter λ =
1 /
6000
= 6000−1
.

(a) Wie groß ist die Wahrscheinlichkeit, dass eine Glühbirne mindestens 6000 Stunden
funktioniert?

(b) Eine Glühbirne sei bereits 3000 Stunden in Betrieb. Wie groß ist die bedingte Wahrscheinlichkeit,dass sie noch weitere 3000 Stunden nicht ausfällt?

4. Sei X eine normalverteilte Zufallsvariable mit µ = 15 und σ = 5. Gesucht ist die Wahrscheinlichkeit,
dass X zwischen 20 und 25 liegt.


```python
from scipy.stats import norm
µ = 15
σ = 5

print(f"P(20 <= X <= 25) = {norm.cdf(25, µ, σ) - norm.cdf(20, µ, σ)}")
```

(d) Wie viel Prozent aller 18-jährigen Studenten sind größer als er als 2m?

8. Die Analyse der Tagesumsätze mittlerer und kleiner Lebensmittelgeschäfte ergab, dass
der Tagesumsatz X (Angaben in e) dieser Geschäfte als eine normalverteilte Zufallsvariable
aufgefasst werden kann, wobei µ = 2000 und σ
2 = 4002 gilt

(a) Wie groß ist die Wahrscheinlichkeit, dass der Tagesumsatz 2500 e übersteigt?

(b) Wie groß ist die Wahrscheinlichkeit, dass der Tagesumsatz zwischen 1600 und 1900
Euro liegt?

(c) Ermitteln Sie das obere Umsatzquartil

(d) Ermitteln Sie den zum Erwartungswert symmetrischen Bereich, in dem der Tagesumsatz
mit einer Wahrscheinlichkeit von 95 % liegt.


```python
µ, σ = 2000, 400
print(f"a) P(x > 2500) = {1-norm.cdf(2500, µ, σ)}")
print(f"b) P(1600 <= x <= 1900) = {norm.cdf(1900, µ, σ) - norm.cdf(1600, µ, σ)}")
print(f"c) {norm.ppf(0.75, µ, σ)}")
untere_grenze = norm.ppf(0.025, µ, σ)
obere_grenze = norm.ppf(0.975, µ, σ)
print(f"d) Symmetrischer 95%-Bereich: [{untere_grenze}, {obere_grenze}]")
```
