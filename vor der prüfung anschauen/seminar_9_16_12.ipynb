1. Für eine Speicherzelle ist die Anzahl X der Betriebsstunden bis zum Auftreten des ersten
Fehlers exponentialverteilt mit dem Parameter λ = 10^−6
.
(a) Wie groß ist die mittlere Zeitdauer bis zum Auftreten des ersten Fehlers?
(b) Mit welcher Wahrscheinlichkeit tritt der erste Fehler innerhalb von 10^6 Betriebsstunden
auf?

```python
from scipy.stats import expon
lam = 10**(-6)
print("(a)  ", 1/lam)
print("(b) ", expon.cdf(10**6, scale=1/lam))
```

Mit welcher Wahrscheinlichkeit warten Sie

(b) länger als 4 Minuten?

(d) weitere 2 Minuten, nachdem Sie schon 3 Minuten gewartet haben
Wie groß ist die Standardabweichung der Wartezeit?

4. Sei X eine normalverteilte Zufallsvariable mit µ = 15 und σ = 5. Gesucht ist die Wahrscheinlichkeit,
dass X zwischen 20 und 25 liegt.

```python
from scipy.stats import norm
mittelwert = 15
standardabweichung = 5

print("P(20<=x<=25) = ", norm.cdf(25, mittelwert, standardabweichung)-norm.cdf(20, mittelwert, standardabweichung))
```

5. Aufgrund von statistischen Erhebungen ist bekannt, dass 18jährige Studenten in etwa
180cm groß sind, und dass die Größe normalverteilt ist mit σ = 7, 5cm.
(a) Definieren Sie die Zufallsvariable
(b) Mit welcher Wahrscheinlichkeit ist ein zufällig ausgewählter 18 jähriger Student
größer als 180cm?
(c) Mit welcher Wahrscheinlichkeit ist er kleiner als 170cm?
(d) Wie viel Prozent aller 18-jährigen Studenten sind größer als er als 2m?
(e) Wie groß muss ein 18-jähriger Student sein, damit nur 5% kleiner sind als er


```python
mittelwert = 180
standardabweichung = 7.5

print("X ~ N(",mittelwert,",", standardabweichung,")")
print("P(x>180) = ", 1-norm.cdf(180, mittelwert, standardabweichung))
print("P(x<170) = ", norm.cdf(170, mittelwert, standardabweichung)-0)
print("P(x>200) = ", 1-norm.cdf(200, mittelwert, standardabweichung))
print("Ein Student muss ", norm.ppf(1-0.05, mittelwert, standardabweichung), " cm groß sein, damit nur 5% kleiner sind als er.")
```

6. Ein Kaffeeautomat kann so eingestellt werden, dass er im Durchschnitt eine bestimmte
Menge Kaffee pro Tasse ausgibt. Die ausgegebenen cl pro Tasse seien normalverteilt mit
σ = 1, 5cl.
Auf welchen Wert µ ist der Automat einzustellen, damit Tassen, die nicht mehr als 25 cl
fassen, nur in 1% der Fälle überfließen?

```python
from scipy.stats import norm

sigma = 1.5
z = norm.ppf(0.99)
mu = 25 - z * sigma

print(mu)
```
