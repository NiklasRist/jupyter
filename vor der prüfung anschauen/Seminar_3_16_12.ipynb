```python
import numpy as np
import matplotlib.pyplot as plt
jahr = np.array([2007, 2008, 2009, 2010, 2011, 2012, 2013])
mio_euro = np.array([53, 65, 72, 87, 100, 107, 130])

plt.scatter(jahr, mio_euro)
# Lineare Regression (Grad 1)
m, b = np.polyfit(jahr, mio_euro, 1)
y_reg = m * jahr + b

# Regressionsgerade
plt.plot(jahr, y_reg, label="Regressionsgerade")
```
