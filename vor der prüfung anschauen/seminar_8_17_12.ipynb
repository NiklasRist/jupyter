2. In der Klausur zur Statistik werden 25 Multiple-Choice-Fragen gestellt mit jeweils vier
Antwortmglichkeiten, von denen genau eine richtige anzukreuzen ist. Wie wahrscheinlich ¨
ist es, mindestens 12 Punkte zu erhalten, wenn man nur rät?

```python
from scipy.stats import binom
n=25
p=1/4
print("P(x>=12) = ", 1-binom.cdf(11, n, p))
```

3. Ein Händler erhält eine umfangreiche Lieferung von Batterien von einem Unternehmen.
Das Unternehmen informiert, dass erfahrungsgemäß 2% der gelieferten Batterien defekt
sind.
Wie groß ist die Wahrscheinlichjeit, dass der Händler in einer Zufallsstichprobe von 40
Batterien höchstens 2 defekte vorfindet?

```python
from scipy.stats import binom
p=0.02
n=40

print("P(x<=2) = ",binom.cdf(2,n, p))
```
