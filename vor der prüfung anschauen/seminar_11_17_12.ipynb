4. Eine Stichprobe von 64 Knopfbatterien für elektrische Kugelschreiber liefert eine mittlere
Lebensdauer von 75 Stunden und eine Standardabweichung von 8 Stunden. Wie groß ist
die mittlere Lebensdauer aller Batterien in einer Lieferung von 1500 Stück mindestens?
(Konfidenzniveau = 0,90)

```python
import scipy.stats as stats
import numpy as np

# Gegebene Werte
x_bar = 75
s = 8
n = 64
confidence = 0.90

alpha = 1 - confidence

# kritischer t-Wert
t_crit = stats.t.ppf(1 - alpha/2, df=n-1)

# Standardfehler
se = s / np.sqrt(n)

# Konfidenzintervall
lower = x_bar - t_crit * se
upper = x_bar + t_crit * se

lower, upper
```
